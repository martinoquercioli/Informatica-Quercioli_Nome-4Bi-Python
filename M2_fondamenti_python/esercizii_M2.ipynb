## Esercizio 1 — Durata di una sessione di laboratorio

*Dati di prova:* `9296`; con `15000` compare anche l'avviso

```python
# Durata di una sessione di laboratorio in ore, minuti e secondi
secondi_totali = int(input("Durata in secondi: "))

if secondi_totali < 0:
    print("la durata non può essere negativa")
else:
    ore = secondi_totali // 3600
    minuti = (secondi_totali % 3600) // 60
    secondi = secondi_totali % 60

    print(f"{ore:02d}:{minuti:02d}:{secondi:02d}")
    print(f"in ore decimali: {secondi_totali / 3600:.2f}")

    if secondi_totali > 4 * 3600:
        print("attenzione: la sessione supera le quattro ore")
```

## Esercizio 2 — Media dei voti e giudizio

*Dati di prova:* `Chiara`, `7.5`, `8`, `6.5`, `12`

```python
# Media di tre voti e giudizio, tenendo conto delle ore di assenza
nome = input("Nome dello studente: ")
voto_1 = float(input("Primo voto: "))
voto_2 = float(input("Secondo voto: "))
voto_3 = float(input("Terzo voto: "))
ore_assenza = int(input("Ore di assenza: "))

media = (voto_1 + voto_2 + voto_3) / 3

if ore_assenza > 50:
    giudizio = "non valutabile"
elif media >= 8:
    giudizio = "ottimo"
elif media >= 6:
    giudizio = "sufficiente"
else:
    giudizio = "insufficiente"

print(f"{nome}: media {media:.2f}, giudizio {giudizio}")
```

## Esercizio 3 — Tavola di conversione incolonnata

*Dati di prova:* `5`, `20`, `5`

```python
# Tavola di conversione da chilometri a miglia e a metri
valore_iniziale = int(input("Valore iniziale (km): "))
valore_finale = int(input("Valore finale (km): "))
passo = int(input("Passo: "))

if passo <= 0:
    print("il passo deve essere positivo: nessuna tabella prodotta")
elif valore_iniziale > valore_finale:
    print("il valore iniziale supera quello finale: nessuna tabella prodotta")
else:
    print(f"{'km':>12}{'miglia':>12}{'metri':>12}")
    print("-" * 36)
    for km in range(valore_iniziale, valore_finale + 1, passo):
        miglia = km * 0.621371
        metri = km * 1000
        print(f"{km:>12}{miglia:>12.3f}{metri:>12}")
```

# Esercizio 5 — Conteggio dei caratteri di una frase

*Dati di prova:* `Il laboratorio 4BI dura 22 ore`

```python
# Conteggio di vocali, consonanti, cifre e altri caratteri di una frase
# Semplificazione voluta: le lettere accentate (à, è, ì, ò, ù) non sono nella
# stringa "aeiou", quindi vengono contate come consonanti.
frase = input("Frase: ")

vocali = 0
consonanti = 0
cifre = 0
altri = 0

for carattere in frase:
    if carattere.isalpha():
        if carattere.lower() in "aeiou":
            vocali += 1
        else:
            consonanti += 1
    elif carattere.isdigit():
        cifre += 1
    else:
        altri += 1

print(f"vocali: {vocali}, consonanti: {consonanti}, cifre: {cifre}, altri: {altri}")

totale = len(frase)
if totale > 0:
    percentuale = vocali / totale * 100
    print(f"percentuale di vocali: {percentuale:.1f}%")
else:
    print("frase vuota: percentuale non calcolabile")
```

## Esercizio 7 — Analisi delle cifre di un numero

*Dati di prova:* `48723`; caso limite `0`

```python
# Numero di cifre, somma delle cifre e cifra maggiore, senza usare stringhe
numero = int(input("Numero intero non negativo: "))

if numero < 0:
    print("il numero deve essere non negativo")
else:
    cifre = 0
    somma = 0
    maggiore = 0
    rimanente = numero

    while rimanente > 0:
        cifra = rimanente % 10
        cifre += 1
        somma += cifra
        if cifra > maggiore:
            maggiore = cifra
        rimanente //= 10

    if numero == 0:
        cifre = 1  # lo zero ha una cifra, ma il ciclo non entra mai

    print(f"cifre: {cifre}")
    print(f"somma delle cifre: {somma}")
    print(f"cifra maggiore: {maggiore}")
```

# Esercizio 8 — Conversione fra decimale e binario nei due sensi

*Dati di prova:* `2026`, poi `11010`

```python
# Da decimale a binario con i resti successivi e da binario a decimale
numero = int(input("Numero intero non negativo: "))
stringa_binaria = input("Stringa di soli zeri e uni: ").strip()

if numero < 0:
    print("il numero deve essere non negativo")
else:
    binario = ""
    rimanente = numero
    while rimanente > 0:
        binario = str(rimanente % 2) + binario
        rimanente //= 2
    if numero == 0:
        binario = "0"

    print(f"{numero} in binario: {binario}")
    if binario == bin(numero)[2:]:
        print("verifica superata")
    else:
        print("verifica fallita")

valida = len(stringa_binaria) > 0
for carattere in stringa_binaria:
    if carattere != "0" and carattere != "1":
        valida = False
        break

if valida:
    valore = 0
    for carattere in stringa_binaria:
        valore = valore * 2 + int(carattere)
    print(f"{stringa_binaria} in decimale: {valore}")
else:
    print("la stringa deve contenere almeno un carattere e solo 0 e 1: nessun calcolo")
```

## Esercizio 9 — Ricerca sequenziale in un elenco di codici

*Dati di prova:* `c307`; seconda esecuzione con `Z999`

```python
# Ricerca sequenziale di un codice in un elenco memorizzato in una stringa
elenco = "A101;B204;C307;D410;E512"
codice = input("Codice da cercare: ").strip().upper()

inizio = 0
posizione = 1
while inizio < len(elenco):
    fine = elenco.find(";", inizio)
    if fine == -1:
        fine = len(elenco)
    voce = elenco[inizio:fine]
    if voce == codice:
        print(f"{codice} è la voce numero {posizione} dell'elenco")
        break
    inizio = fine + 1
    posizione += 1
else:
    print(f"{codice} non è presente nell'elenco")
```

## Esercizio 10 — Statistiche di una sequenza letta con sentinella

*Dati di prova:* `7`, `5.5`, `15`, `9`, `6`, `8`, `0`

```python
# Statistiche su una sequenza di voti terminata dalla sentinella 0
validi = 0
somma = 0.0
massimo = 0.0
minimo = 0.0
sufficienze = 0

voto = float(input("Voto (0 per terminare): "))
while voto != 0:
    if voto < 1 or voto > 10:
        print(f"valore {voto} fuori intervallo: ignorato")
    else:
        validi += 1
        somma += voto
        if validi == 1:
            # massimo e minimo partono dal primo valore valido
            massimo = voto
            minimo = voto
        else:
            if voto > massimo:
                massimo = voto
            if voto < minimo:
                minimo = voto
        if voto >= 6:
            sufficienze += 1
    voto = float(input("Voto (0 per terminare): "))

if validi == 0:
    print("nessun voto valido inserito: statistiche non calcolate")
else:
    print(f"voti validi: {validi}")
    print(f"massimo: {massimo:.2f}")
    print(f"minimo: {minimo:.2f}")
    print(f"media: {somma / validi:.2f}")
    print(f"sufficienze: {sufficienze}")
```

## Esercizio 11 — Palindroma con i due indici

*Dati di prova:* `I topi non avevano nipoti`

```python
# Verifica di una frase palindroma con due indici che si avvicinano
frase = input("Frase: ")

ripulita = ""
for carattere in frase:
    if carattere.isalpha():
        ripulita += carattere.lower()

print(f"testo normalizzato: {ripulita}")

if len(ripulita) == 0:
    print("la frase non contiene lettere: nulla da verificare")
else:
    sinistra = 0
    destra = len(ripulita) - 1
    confronti = 0
    palindroma = True

    while sinistra < destra:
        confronti += 1
        if ripulita[sinistra] != ripulita[destra]:
            palindroma = False
            break
        sinistra += 1
        destra -= 1

    print(f"palindroma: {palindroma}")
    print(f"confronti eseguiti: {confronti}")
    print(f"controllo con lo slicing: {ripulita == ripulita[::-1]}")
```

## Esercizio 12 — Occorrenze di una sottostringa senza il metodo count

*Dati di prova:* `la lampada della lavagna`, poi `la`

```python
# Occorrenze di una sottostringa cercate con find, senza count e senza replace
testo = input("Testo: ")
sottostringa = input("Sottostringa da cercare: ")

if len(sottostringa) == 0:
    print("la sottostringa da cercare non può essere vuota")
else:
    occorrenze = 0
    posizioni = ""
    mascherato = ""
    ultimo = 0

    posizione = testo.find(sottostringa)
    while posizione != -1:
        occorrenze += 1
        posizioni += str(posizione) + " "
        mascherato += testo[ultimo:posizione] + "*" * len(sottostringa)
        ultimo = posizione + len(sottostringa)
        posizione = testo.find(sottostringa, ultimo)
    mascherato += testo[ultimo:]

    print(f"occorrenze: {occorrenze}")
    if occorrenze == 0:
        print("posizioni: nessuna")
    else:
        print(f"posizioni: {posizioni.strip()}")
    print(f"verifica con count: {testo.count(sottostringa)}")
    print(f"testo mascherato: {mascherato}")

```
