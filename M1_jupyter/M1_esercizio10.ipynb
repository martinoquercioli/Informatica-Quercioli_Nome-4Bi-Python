# Procedura di consegna di un esercizio

**Destinatario:** Martino Quercioli

Questo documento descrive come consegnare un esercizio nel repository. Ogni passo esiste per mantenere verificabile il lavoro; saltare un controllo può lasciare file errati o non tracciati.

````python
passi_previsti = 5
print("Passi previsti:", passi_previsti)
````

## 1. Modifica il file

Modifica il documento richiesto e conserva il percorso indicato dalla traccia. Se cambi nome o cartella, il file può non essere trovato durante la consegna.

```bash
code M1_markdown/esercizio.md
```

````python
minuti_per_passo = 5
tempo_stimato = passi_previsti * minuti_per_passo
print("Tempo stimato:", tempo_stimato, "minuti")
````

## 2. Controlla l'anteprima

Apri l'anteprima Markdown e controlla titoli, paragrafi, elenchi e recinti. Se salti questa verifica, un errore di sintassi può restare invisibile fino alla correzione.

```bash
git diff --check
```

````python
file_nel_commit = 1
print("File nel commit:", file_nel_commit)
````

## 3. Controlla lo stato

Esegui `git status` per vedere i file modificati e non tracciati. Senza questo controllo potresti lasciare fuori la consegna oppure aggiungere file temporanei.

```bash
git status
```

| Comando | Effetto |
|---|---|
| `git diff --check` | Cerca errori di spaziatura |
| `git status` | Mostra lo stato del repository |
| `git add` | Prepara i file al commit |
| `git commit` | Registra una versione |

## 4. Crea il commit

Aggiungi solo i file dell'esercizio e usa un messaggio conforme alla convenzione. Se salti il commit, il lavoro resta soltanto nella copia locale.

```bash
git add M1_markdown/esercizio.md
git commit -m "Aggiunge esercizio (M1 es. 1)"
```

## 5. Verifica la consegna

Controlla il commit e il rendering sul repository remoto quando la traccia lo richiede. Se salti l'ultima verifica, puoi consegnare un file diverso da quello controllato localmente.

> Prima della consegna controlla l'anteprima e, per i notebook, riavvia il kernel ed esegui tutte le celle.
